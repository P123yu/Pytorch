```python
df = pd.read_csv("/kaggle/input/datasets/abhishek14398/salary-dataset-simple-linear-regression/Salary_dataset.csv")
df.head()
```

```python
df = df.drop(columns=['Unnamed: 0'])
df.head()
```

```python
X , y = df['YearsExperience'],df['Salary']
print(X[:2])
print(y[:2])
```

```python
X_data = np.array(X).reshape(-1,1)
y_data = np.array(y).reshape(-1,1)
print(X_data[:2])
print(y_data[:2])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train_data, X_test_data, y_train_data,y_test_data = train_test_split(X_data,y_data,test_size=0.2,random_state=42)

X_scale = StandardScaler()
y_scale = StandardScaler() 

X_train_scaled = X_scale.fit_transform(X_train_data)
y_train_scaled = y_scale.fit_transform(y_train_data)

X_test_scaled = X_scale.transform(X_test_data)
y_test_scaled = y_scale.transform(y_test_data)

print(X_train_scaled[:2])
print(X_test_scaled[:2])
```

```python
import torch

X_train_scaled_torch = torch.tensor(X_train_scaled, dtype = torch.float32)
X_test_scaled_torch = torch.tensor(X_test_scaled, dtype = torch.float32)
y_train_scaled_torch = torch.tensor(y_train_scaled, dtype = torch.float32)
y_test_scaled_torch = torch.tensor(y_test_scaled, dtype = torch.float32)

print(X_train_scaled_torch[:2])
print(X_test_scaled_torch[:2])
```

```python
import torch.nn as nn
import torch.optim as optim 

model = nn.Linear(1,1)
loss_fun = nn.MSELoss()
optimizer = optim.SGD(model.parameters(),lr=0.01)

epochs = 500
for epoch in range(epochs):
    model.train()
    y_trained_output = model(X_train_scaled_torch)
    loss = loss_fun(y_trained_output,y_train_scaled_torch)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    if epoch % 100 == 0:
        print(loss.item())



from sklearn.metrics import r2_score,mean_absolute_error
model.eval()
with torch.no_grad():
    y_test_output = model(X_test_scaled_torch)
    r2 = r2_score(y_test_scaled_torch,y_test_output)
    mae = mean_absolute_error(y_test_scaled_torch,y_test_output)
    print(f"r2_score: {r2}")
    print(f"mae: {mae}")
```
